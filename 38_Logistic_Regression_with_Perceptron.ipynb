<a id='step'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 5</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Activation Functions � Step vs Sigmoid</h2>
</div>

The choice of activation function is the key difference between the Perceptron and Logistic Regression. Both use the same linear combination `z = w�x + b` � only what they do with `z` differs.

```python
def step(z):
    """Heaviside step function � hard binary output."""
    return 1 if z >= 0 else 0

def sigmoid(z):
    """Sigmoid (logistic) function � smooth probability output."""
    return 1 / (1 + np.exp(-z))

z_vals   = np.linspace(-6, 6, 300)
step_out = np.array([step(z) for z in z_vals])
sig_out  = sigmoid(z_vals)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Step
axes[0].plot(z_vals, step_out, color='#00d4ff', linewidth=2.5)
axes[0].axvline(0, color='gray', linewidth=0.8, linestyle='--')
axes[0].axhline(0.5, color='gray', linewidth=0.8, linestyle='--')
axes[0].fill_between(z_vals, step_out, alpha=0.1, color='#00d4ff')
axes[0].set_title('Step Function � Perceptron', fontsize=12, fontweight='bold')
axes[0].set_xlabel('z = w�x + b', fontsize=11)
axes[0].set_ylabel('Output', fontsize=11)
axes[0].set_yticks([0, 0.5, 1])
axes[0].annotate('Hard jump at z=0\nNo gradient here!', xy=(0, 0.5),
                 xytext=(1.5, 0.3), fontsize=9, color='#e74c3c',
                 arrowprops=dict(arrowstyle='->', color='#e74c3c'))
axes[0].grid(True, alpha=0.3)

# Sigmoid
axes[1].plot(z_vals, sig_out, color='#f0a500', linewidth=2.5)
axes[1].axvline(0, color='gray', linewidth=0.8, linestyle='--')
axes[1].axhline(0.5, color='gray', linewidth=0.8, linestyle='--')
axes[1].fill_between(z_vals, sig_out, alpha=0.1, color='#f0a500')
axes[1].set_title('Sigmoid Function � Logistic Regression', fontsize=12, fontweight='bold')
axes[1].set_xlabel('z = w�x + b', fontsize=11)
axes[1].set_ylabel('s(z) � Probability', fontsize=11)
axes[1].set_yticks([0, 0.5, 1])
axes[1].annotate('Smooth curve\nGradient exists everywhere', xy=(0, 0.5),
                 xytext=(1.5, 0.25), fontsize=9, color='#2ecc71',
                 arrowprops=dict(arrowstyle='->', color='#2ecc71'))
axes[1].grid(True, alpha=0.3)

plt.suptitle('Step vs Sigmoid � Hard vs Soft Activation', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

<div style="background: #0d1f2d; border-left: 4px solid #00d4ff; border-radius: 0 8px 8px 0; padding: 14px 18px; margin: 12px 0; font-size: 14px; color: #7ecfdf;">
  <strong style="color: #e6f4f8;">?? Why the step function limits learning:</strong> The step function has <strong>zero gradient almost everywhere</strong> (and undefined gradient at z=0). This means gradient-based optimisation is impossible � the Perceptron can only learn through the error-correction rule, not by following a loss landscape.
</div>

<a id='perceptron'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #e74c3c; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Approach 1</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">?? Perceptron � Step Function from Scratch</h2>
</div>

We implement the classic Perceptron: picks one random sample per iteration, predicts using the step function, and updates weights only when a mistake is made.

```python
def perceptron_step(X, y, lr=0.01, epochs=1000):
    """
    Classic Perceptron with hard step function activation.
    Updates weights only on misclassified samples.
    """
    y = np.asarray(y).ravel()
    X_b     = np.insert(X, 0, 1, axis=1)     # prepend bias
    weights = np.ones(X_b.shape[1])

    for _ in range(epochs):
        j     = np.random.randint(0, X_b.shape[0])
        y_hat = step(np.dot(X_b[j], weights))
        weights += lr * (y[j] - y_hat) * X_b[j]

    return weights[0], weights[1:]             # intercept, coef

intercept_step, coef_step = perceptron_step(X, y, lr=0.01, epochs=1000)

print(f"Perceptron (step) coefficients : {coef_step}")
print(f"Perceptron (step) intercept    : {intercept_step:.4f}")
```

```python
# Derive slope & intercept for the decision boundary line: x2 = m*x1 + b
m_step = -(coef_step[0] / coef_step[1])
b_step = -(intercept_step / coef_step[1])

x_line = np.linspace(-3, 3, 200)

plt.figure(figsize=(9, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100, edgecolors='white', linewidth=0.5, zorder=3)
plt.plot(x_line, m_step * x_line + b_step, color='red', linewidth=2.5,
         label=f'Perceptron (step) boundary')
plt.xlabel('Feature 1', fontsize=12)
plt.ylabel('Feature 2', fontsize=12)
plt.title('Perceptron � Step Function Decision Boundary', fontsize=13, fontweight='bold')
plt.legend(fontsize=11)
plt.ylim(-3, 3)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

<a id='anim1'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 7</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Animating the Perceptron Learning Process</h2>
</div>

The animation below shows the decision boundary **rotating and shifting** with each epoch as the Perceptron corrects its mistakes. We record the boundary at every step, then replay it.

```python
def perceptron_animated(X, y, lr=0.1, epochs=200):
    """Perceptron that records the boundary at every epoch for animation."""
    y = np.asarray(y).ravel()
    slopes, intercepts = [], []

    X_b     = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X_b.shape[1])

    for _ in range(epochs):
        j     = np.random.randint(0, X_b.shape[0])
        y_hat = step(np.dot(X_b[j], weights))
        weights += lr * (y[j] - y_hat) * X_b[j]

        if weights[2] != 0:
            slopes.append(-(weights[1] / weights[2]))
            intercepts.append(-(weights[0] / weights[2]))
        else:
            slopes.append(slopes[-1] if slopes else 0)
            intercepts.append(intercepts[-1] if intercepts else 0)

    return slopes, intercepts

m_hist, b_hist = perceptron_animated(X, y, lr=0.1, epochs=200)

x_anim = np.arange(-3, 3, 0.1)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(X[:, 0], X[:, 1], c=np.asarray(y).ravel(),
           cmap='winter', s=100, edgecolors='white', linewidth=0.5, zorder=3)
line, = ax.plot(x_anim, x_anim * m_hist[0] + b_hist[0], 'r-', linewidth=2.5, zorder=2)
ax.set_ylabel('Feature 2', fontsize=11)
ax.set_title('Perceptron � Decision Boundary Learning', fontsize=12, fontweight='bold')
ax.set_ylim(-3, 3)
ax.grid(True, alpha=0.3)

def update(i):
    line.set_ydata(x_anim * m_hist[i] + b_hist[i])
    ax.set_xlabel(f'Epoch {i + 1} / 200', fontsize=11)

anim = FuncAnimation(fig, update, frames=200, interval=80, repeat=True)
plt.tight_layout()
plt.close()
HTML(anim.to_jshtml())
```

We fit sklearn's `LogisticRegression` once here and treat it as the **reference solution** throughout the notebook. Every scratch implementation will be compared against this.

```python
lor = LogisticRegression()
lor.fit(X, y)

# Decision boundary line for sklearn
m_lor = -(lor.coef_[0][0] / lor.coef_[0][1])
b_lor = -(lor.intercept_[0]  / lor.coef_[0][1])

print(f"sklearn coefficients : {lor.coef_[0]}")
print(f"sklearn intercept    : {lor.intercept_[0]:.4f}")
print(f"sklearn accuracy     : {accuracy_score(y, lor.predict(X)):.4f}")
print(f"Boundary slope m     : {m_lor:.3f}")
print(f"Boundary intercept b : {b_lor:.3f}")
```

  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Comparison 1 � Perceptron (Step) vs sklearn</h2>

<div style="background: #1a2a1a; border-left: 4px solid #f39c12; border-radius: 0 8px 8px 0; padding: 14px 18px; margin: 12px 0; font-size: 14px;">
  <strong style="color: #f39c12;">?? Problem with the Step Perceptron:</strong> <span style="color: #a0c8a0;">The red dashed boundary doesn't align with sklearn's black boundary. The Perceptron finds <em>any</em> separating hyperplane � not necessarily the optimal one. The step function gives no gradient signal to guide the weights toward the probabilistically best boundary.</span>
</div>

<a id='sigmoid-perceptron'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #f0a500; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Approach 2</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">?? Sigmoid Perceptron � Soft Outputs, Same Update Rule</h2>
</div>

The simplest upgrade: replace the hard step function with the **sigmoid**. Now `y_hat` is a probability (0�1) instead of a binary label. The same weight update rule applies � but the error signal `(y - y)` is now continuous, giving smoother and more informative updates.

<div style="background: #1a1a0d; border: 1px solid #3a3a00; border-radius: 8px; padding: 12px 18px; margin: 14px 0; font-size: 13px; color: #c8c870;">
  <strong style="color: #f0a500;">?? What changes:</strong> <code style='color:#f0a500;'>y_hat = step(z)</code> ? <code style='color:#f0a500;'>y_hat = sigmoid(z)</code>. Everything else stays the same � same random sample selection, same update rule, same one-sample-at-a-time approach.
</div>

```python
def perceptron_sigmoid(X, y, lr=0.01, epochs=1000):
    """
    Perceptron with sigmoid activation instead of hard step.
    Produces probabilistic outputs � smoother error signal.
    Still stochastic: one random sample per update.
    """
    y = np.asarray(y).ravel()
    X_b     = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X_b.shape[1])

    for _ in range(epochs):
        j     = np.random.randint(0, X_b.shape[0])
        y_hat = sigmoid(np.dot(X_b[j], weights))    # ? sigmoid, not step
        weights += lr * (y[j] - y_hat) * X_b[j]

    return weights[0], weights[1:]

intercept_sig, coef_sig = perceptron_sigmoid(X, y, lr=0.01, epochs=1000)

m_sig = -(coef_sig[0] / coef_sig[1])
b_sig = -(intercept_sig / coef_sig[1])

print(f"Sigmoid Perceptron coefficients : {coef_sig}")
print(f"Sigmoid Perceptron intercept    : {intercept_sig:.4f}")
print(f"Decision boundary: x2 = {m_sig:.3f}�x1 + {b_sig:.3f}")
```

<a id='compare2'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 11</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Comparison 2 � Perceptron vs Sigmoid vs sklearn</h2>
</div>

```python
plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100,
            edgecolors='white', linewidth=0.5, zorder=3, label='Data')
plt.plot(x_plot, m_step * x_plot + b_step,
         color='red',     linewidth=2,   linestyle='--',  label='Perceptron (step)')
plt.plot(x_plot, m_sig  * x_plot + b_sig,
         color='#f0a500', linewidth=2,   linestyle='-.',   label='Sigmoid Perceptron')
plt.plot(x_plot, m_lor  * x_plot + b_lor,
         color='black',   linewidth=2.5,                   label='sklearn LogisticRegression')
plt.xlabel('Feature 1', fontsize=12)
plt.ylabel('Feature 2', fontsize=12)
plt.title('Step vs Sigmoid Perceptron vs sklearn', fontsize=13, fontweight='bold')
plt.legend(fontsize=11)
plt.ylim(-3, 3)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

<div style="background: #1a1a0d; border-left: 4px solid #f0a500; border-radius: 0 8px 8px 0; padding: 14px 18px; margin: 12px 0; font-size: 14px;">
  <strong style="color: #f0a500;">?? Problem with the Sigmoid Perceptron:</strong> <span style="color: #c8c870;">Better than the step function, but the boundary still doesn't match sklearn's. Why? Because we're still updating on <strong>one random sample at a time</strong> � the updates are noisy and don't follow the true gradient of the loss function over the entire dataset. We need <strong>batch gradient descent</strong>.</span>
</div>

<a id='gd'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Approach 3</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">?? Logistic Regression via Batch Gradient Descent</h2>
</div>

The fix: compute the gradient over the **entire dataset** at every step. This is the true gradient of binary cross-entropy loss, and it's exactly what sklearn optimises internally.

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 12px; padding: 22px 28px; margin: 16px 0;">
  <div style="color: #7ecfdf; font-size: 11px; letter-spacing: 1.5px; text-transform: uppercase; margin-bottom: 14px;">Log Loss Gradient � Full Batch</div>
  <div style="display: flex; flex-direction: column; gap: 10px;">
    <div>
      <span style="color: #7ecfdf; font-size: 12px;">? Predictions for ALL samples at once</span><br/>
      <span style="color: #e0f0f8; font-size: 15px; font-family: Georgia, serif;">y = s(Xw)</span>
    </div>
    <div>
      <span style="color: #7ecfdf; font-size: 12px;">? Gradient of log loss w.r.t. weights</span><br/>
      <span style="color: #00d4ff; font-size: 15px; font-family: Georgia, serif;">?L = X?(y - y) / n &nbsp;&nbsp; ? &nbsp;&nbsp; use (y - y) to ascend</span>
    </div>
    <div>
      <span style="color: #7ecfdf; font-size: 12px;">? Update weights</span><br/>
      <span style="color: #e0f0f8; font-size: 15px; font-family: Georgia, serif;">w ? w + ? � X?(y - y) / n</span>
    </div>
  </div>
  <div style="color: #7ecfdf; font-size: 12px; margin-top: 12px; border-top: 1px solid #1e4d6b; padding-top: 10px;">
    The key difference from the perceptron: <code>np.dot(X.T, y - y_hat)</code> sums the error signal across <strong>all samples simultaneously</strong> � not just one random sample.
  </div>
</div>

```python
def logistic_regression_gd(X, y, lr=0.5, epochs=5000):
    """
    Logistic Regression via Batch Gradient Descent.
    Minimises binary cross-entropy using the full dataset at every step.
    This is the correct implementation that matches sklearn's solution.

    Update rule:  w ? w + ? � X?(y - s(Xw)) / n
    """
    y = np.asarray(y).ravel()
    X_b     = np.insert(X, 0, 1, axis=1)      # prepend bias column
    weights = np.ones(X_b.shape[1])
    n       = X_b.shape[0]

    for _ in range(epochs):
        y_hat   = sigmoid(np.dot(X_b, weights))           # all samples at once
        weights += lr * (np.dot(X_b.T, (y - y_hat)) / n) # batch gradient update

    return weights[1:], weights[0]             # coef, intercept

coef_gd, intercept_gd = logistic_regression_gd(X, y, lr=0.5, epochs=5000)

m_gd = -(coef_gd[0] / coef_gd[1])
b_gd = -(intercept_gd / coef_gd[1])

print(f"GD coefficients : {coef_gd}")
print(f"GD intercept    : {intercept_gd:.4f}")
print(f"Decision boundary: x2 = {m_gd:.3f}�x1 + {b_gd:.3f}")
print()
print(f"sklearn coef     : {lor.coef_[0]}")
print(f"sklearn intercept: {lor.intercept_[0]:.4f}")
```
